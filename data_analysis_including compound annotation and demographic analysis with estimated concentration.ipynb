```python
import pandas as pd

# Combine isotopic pattern scores with compound annotation and detection frequency

# Isotopic pattern scoring table
isotopic_pattern = pd.read_csv(r"D:\UCSF_postdoc_topic\Collaboration\prenatal_exposure_QAC\matched_peak_pos_with_qac_isotope_scores.csv")
print(isotopic_pattern.columns)

# Detection frequency table after blank subtraction
DF_afterbksub = pd.read_csv(r"D:\UCSF_postdoc_topic\Collaboration\prenatal_exposure_QAC\Manuscript_draft\corrected_average_peak_area_pos_QACs_0505.csv")
print(DF_afterbksub.columns)

# Final annotation table
annotation_table = pd.read_csv(r"D:\UCSF_postdoc_topic\Collaboration\prenatal_exposure_QAC\Manuscript_draft\Annotation_table.csv")
print(annotation_table.columns)


# Retrieve isotopic combined score by matching DTXSID and feature_id
annotation_table = annotation_table.merge(
    isotopic_pattern[
        ["DTXSID", "feature_id", 'qac_isotope_cosine_score', "qac_combined_ms1_score"]
    ],
    on=["DTXSID", "feature_id"],
    how="left"
)


# Retrieve detection frequency after blank subtraction by matching feature_id
annotation_table = annotation_table.merge(
    DF_afterbksub[
        ["feature_id", "Detection frequency"]
    ],
    on="feature_id",
    how="left"
)


# Optional: save updated annotation table
annotation_table.to_csv(
    r"D:\UCSF_postdoc_topic\Collaboration\prenatal_exposure_QAC\Manuscript_draft\Annotation_table_with_MS1_scores_and_DF_final.csv",
    index=False
)
```
